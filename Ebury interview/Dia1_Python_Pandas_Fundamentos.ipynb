### Tuplas y sets (más breve, pero conviene reconocerlos)

- **Set**: colección sin duplicados y sin orden — útil para saber, por ejemplo, cuántas divisas distintas hay en un dataset.

```python
lista_con_repetidos = ["USD", "EUR", "USD", "GBP", "EUR"]
divisas_unicas = set(lista_con_repetidos)
print(divisas_unicas)
print(len(divisas_unicas))   # cuántas divisas distintas hay
```

## Parte 2 — Funciones

Repaso rápido de sintaxis, con foco en lo que probablemente necesites: funciones con valores por defecto, y `*args`/`**kwargs` (menos crítico, pero puede aparecer).

```python
def convertir_a_eur(importe, divisa, tipos_cambio, base="EUR"):
    """Convierte un importe desde su divisa original a EUR usando un diccionario de tipos de cambio."""
    if divisa == base:
        return importe
    return importe / tipos_cambio[divisa]

tipos_cambio = {"USD": 1.08, "GBP": 0.86}
print(convertir_a_eur(100000, "USD", tipos_cambio))
print(convertir_a_eur(50000, "EUR", tipos_cambio))
```

### Selección: `.loc` vs `.iloc`
- `.loc` selecciona por **etiqueta** (nombre de columna, valor de índice).
- `.iloc` selecciona por **posición** (como un índice numérico de lista).

Esta es una de las confusiones más típicas — practícala hasta que sea automática.

```python
print(df.loc[0, "divisa"])        # fila con índice 0, columna "divisa"
print(df.iloc[0, 1])              # primera fila, segunda columna (posición)

print(df.loc[0:2])                # filas con índice 0,1,2 (incluye el final con .loc)
print(df.iloc[0:2])               # filas en posición 0,1 (excluye el final, como slicing normal)
```

```python
# TODO: inténtalo con un enfoque "manual" (bucle + filtrado), sin usar groupby todavía

# Method 1:
def resumen_por_divisa_manual(df):
    unique_curr=df["divisa"].unique()
    sum_unique_curr=np.zeros(len(unique_curr))
    transactions=np.zeros(len(unique_curr))
    for i in range(len(unique_curr)):
        sum_unique_curr[i]=sum(df["importe"][df["divisa"]==unique_curr[i]])
        transactions[i]=len(df[df["divisa"]==unique_curr[i]])
    summary=pd.DataFrame(unique_curr,columns=["divisa"])
    summary["Sum"]=sum_unique_curr
    summary["Transactions"]=transactions

    return summary
print(resumen_por_divisa_manual(df))

# Method 2:
def resumen_por_divisa_manual2(df):
    final_summary=[]
    for i in df["divisa"].unique():
        subset=df[df["divisa"]==i]
        suma=sum(subset["importe"])
        trans=len(subset)
        final_summary.append({
            "Currency": i,
            "Sum":suma,
            "Transactions":trans
        })
        final_summary_df=pd.DataFrame(final_summary)
    return final_summary_df

print(resumen_por_divisa_manual2(df))
```

**Solución 3.2 (enfoque manual, sin groupby)**

```python
def resumen_por_divisa_manual(df):
    filas = []
    for divisa in df["divisa"].unique():
        subset = df[df["divisa"] == divisa]
        filas.append({
            "divisa": divisa,
            "importe_total": subset["importe"].sum(),
            "n_transacciones": len(subset)
        })
    return pd.DataFrame(filas)

print(resumen_por_divisa_manual(df))
```

### El error más típico bajo presión: `SettingWithCopyWarning`
Esto le pasa a todo el mundo alguna vez, y en una entrevista en vivo puede ponerte nervioso si no sabes qué significa. Ocurre cuando modificas un DataFrame que en realidad es una "vista" (un subconjunto filtrado) de otro, y pandas no está seguro de si debe modificar el original o no.

```python
# Esto puede lanzar un SettingWithCopyWarning:
usd_mal = df[df["divisa"] == "USD"]
# usd_mal["revisado"] = True   # <- esta línea generaría el warning si la descomentas

# La forma correcta: usa .copy() explícitamente cuando vayas a modificar un subconjunto
usd_bien = df[df["divisa"] == "USD"].copy()
usd_bien["revisado"] = True
print(usd_bien[["divisa", "revisado"]])
```

> **Regla práctica para el lunes**: si vas a filtrar un DataFrame y luego modificarlo (añadir columnas, cambiar valores), añade `.copy()` justo después del filtrado. Te ahorra el warning y, más importante, te ahorra el bug real de modificar sin querer el DataFrame original.
