### Comprensiones en listas y diccionarios

Let´s replicate how to **compress** data in an efficient way

```python
importes=[1000, -500, 2300, -150, 800]
positivos_bucle={}

# Method 1: with indexes
method=0
positivos_bucle[method]=[]
for i in range(len(importes)):
    if importes[i]>0:
        positivos_bucle[method].append(importes[i])
print(positivos_bucle)

# Method 2: with compressio
method=1
positivos_bucle[method]=[]

for x in importes:
    if x>0:
        positivos_bucle[method].append(x)
print(positivos_bucle)

# Method 3: final compression
method=2
positivos_bucle[method]=[]

positivos_bucle[method]=[x for x in importes if x>0]
print(positivos_bucle)
```
