### Exercise 8 (Python basics)

```python
fx_rates = {"USD":1.08,"GBP":0.86}
# TODO
def to_eur(amount, currency):
    pass

print(to_eur(1000,"USD"))
print(to_eur(500,"EUR"))
print(to_eur(200,"CHF"))
```

**Solution 8**

```python
fx_rates = {"USD":1.08,"GBP":0.86}
def to_eur(amount, currency):
    if currency in fx_rates:
        return amount / fx_rates[currency]
    return amount

print(to_eur(1000,"USD"))
print(to_eur(500,"EUR"))
print(to_eur(200,"CHF"))
```

### Exercise 20 (SQL with CTE)

```python
# TODO
cte_query = """

"""
# print(duckdb.sql(cte_query).df())
```

**Solution 20**

```python
cte_query = """
WITH totals AS (
  SELECT currency, SUM(amount) as total
  FROM df
  GROUP BY currency
)
SELECT * FROM totals WHERE total > 0
"""
print(duckdb.sql(cte_query).df())
```
